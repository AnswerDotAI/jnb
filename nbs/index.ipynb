# jnb

> The J language in notebooks, with j magics for IPython and Jupyter and a kernmini-based J kernel

`jnb` runs the [J language](https://www.jsoftware.com/) in notebooks. It provides `%%j` and `%j` magics for Python notebooks, and a Jupyter kernel for notebooks written in J. Both load libj, J's engine library, into the Python process.

## Installation

Install J with `pip install jlanguage` or from [jsoftware.com](https://www.jsoftware.com/). Then install jnb:

    pip install jnb

To register the magics automatically in IPython and Jupyter sessions, run:

    jnb_install

You can also load them for one session with `%load_ext jnb`. The J engine starts when you first run a magic.

## The `j` magics

`%%j` runs code and displays J's output verbatim. `%j expr` returns an expression's value as Python data. The session keeps nouns and verbs across cells:

```python
#| hide
%load_ext jnb
```

```python
%%j
m =: 3 3 $ i. 9
m +/ . * m
```

```python
z = %j m
z
```

You can also use the `J` session class directly. It runs J code, transfers values in both directions with `[]`, and converts J verbs to Python callables with `fn`. See the [`j` module documentation](https://answerdotai.github.io/jnb/j.html) for examples.

## The J Jupyter kernel

`jkernel` is a Jupyter kernel for J built on [kernmini](https://github.com/AnswerDotAI/kernmini). Register it with:

    python -m jnb.jkernel install

Use it from JupyterLab, nbclient, [jupygate](https://github.com/AnswerDotAI/jupygate), or agents using clikernel. Workspace state persists across cells. A J attention interrupt stops a computation without resetting the workspace.

The kernel runs `jnb/startup.ijs` from your XDG configuration directory when that file exists. The default path is `~/.config/jnb/startup.ijs`.


## Limitations

- `%j` can return booleans, integers, floats and characters as Python data. Boxed, extended and rational values raise an error. Use `%%j` to display them.
- macOS and Linux are supported. The libj binding is untested on Windows, which is not yet supported.
